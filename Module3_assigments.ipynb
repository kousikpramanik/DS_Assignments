2.Describe the role of predefined keywords in Python and provide examples of how they are used in a
program

```python
Predefined keywords in Python are reserved words that have special meanings and cannot be used as variable names, function names, or identifiers. They define the syntax and structure of the language and help in controlling the flow of execution. These keywords are essential for writing logical and structured programs in Python.

Python keywords serve different purposes, such as defining control flow (if, else, for, while), handling exceptions (try, except, finally), defining functions (def, return), and working with data types (True, False, None).

Examples:

x = 10
if x > 5:
    print("x is greater than 5")
elif x == 5:
    print("x is equal to 5")
else:
    print("x is less than 5")

```

3. Compare and contrast mutable and immutable objects in Python with examples.


```python
#Mutable objects can be modified after creation where as immutable object can't be modified after its creation. List,dictionaries, sets are the example of mutable objects and Strings, tuples are the example of immutable objects in python.

Exampls:

#mutable

my_list = [1, 2, 3]
print(id(my_list))  # Memory address before modification

my_list.append(4)  # Modifying the list
print(my_list)  # Output: [1, 2, 3, 4]
print(id(my_list))  # Same memory address (list modified in place)

#immutable

s = "Hello"
print(id(s))  # Memory address before modification

s = s + " World"  # Creates a new string
print(s)  # Output: Hello World
print(id(s))  # Different memory address (new object created)

```
